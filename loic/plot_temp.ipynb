```python
import xarray as xr
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
from rasterio.features import geometry_mask
from rasterio.transform import from_bounds
from pathlib import Path

# Charger le champ ERA5
db_file = Path('../tests/era5_t2m_20250101_fr.nc')
ds = xr.open_dataset(db_file)
da = ds['t2m'].isel(valid_time=0)

# Convertir en °C si nécessaire
units = da.attrs.get('units', '')
if units.lower() in {'k', 'kelvin'}:
    da = da - 273.15
    units = '°C'

# Harmoniser lon/lat : garder latitude décroissante (Nord -> Sud) pour aligner le masque rasterio
if float(da.longitude.max()) > 180:
    da = da.assign_coords(longitude=((da.longitude + 180) % 360) - 180).sortby('longitude')
if float(da.latitude[0]) < float(da.latitude[-1]):
    da = da.sortby('latitude', ascending=False)

lats = da['latitude'].values
lons = da['longitude'].values

# Contours France (régions + départements)
shapes_dir = Path('../data/shapefiles')
regions = gpd.read_file(shapes_dir / 'region.shp').to_crs('EPSG:4326')
departements = gpd.read_file(shapes_dir / 'departement.shp').to_crs('EPSG:4326')
france_geom = regions.unary_union

# Masque rasterisé sur la grille ERA5
step_x = float(np.abs(np.diff(lons)).mean())
step_y = float(np.abs(np.diff(lats)).mean())
west = float(lons.min() - step_x / 2)
east = float(lons.max() + step_x / 2)
south = float(lats.min() - step_y / 2)
north = float(lats.max() + step_y / 2)
transform = from_bounds(west, south, east, north, len(lons), len(lats))
mask = geometry_mask([france_geom], out_shape=(len(lats), len(lons)), transform=transform, invert=True, all_touched=False)

# Température masquée à la France

da_france = da.where(mask)

# Plot température + frontières (légende + figure ajustées)
fig, ax = plt.subplots(figsize=(9, 9))
cb = da_france.plot(
    ax=ax,
    x='longitude',
    y='latitude',
    add_colorbar=True,
    cmap='coolwarm',
    cbar_kwargs={'shrink': 0.6, 'pad': 0.02}
)
regions.boundary.plot(ax=ax, linewidth=0.8, edgecolor='black', zorder=2)
departements.boundary.plot(ax=ax, linewidth=0.3, edgecolor='dimgray', zorder=2)

ax.set_title(f"Température 2m masquée France [{units}]")
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_aspect('equal', adjustable='box')
plt.show()
```
